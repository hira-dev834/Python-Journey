```python
text = "Hira"
number = 19

# Print fuction is suitable for both data types.
print(text)
print(number)

# Type fuction shows the string and integer type of both.
print(type(text))
print(type(number))

# Length fuction works for only string data type.
print(len(text))

#Upper method only works for string data type.
print(text.upper())

#Bit length method only works for integer.
print(number.bit_length())
```
